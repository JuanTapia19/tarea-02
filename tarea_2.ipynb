```python
# Hacemos un bucle para recorrer los números desde -30 hasta 50.
# Usamos 51 porque la función range() siempre corta un número antes del final.
for temp in range(-30, 51):

    # Primera condición: si la temperatura es menor a cero
    if temp < 0:
        print(f"{temp} Muy frío")

    # Segunda condición: si está entre 0 y 24.
    # Como el 'if' anterior ya filtró los negativos, aquí solo llegan los mayores o iguales a 0.
    elif temp <= 24:
        print(f"{temp} Templado")

    # Última condición: si no fue menor a 0 ni menor o igual a 24,
    # por descarte lógico es 25 o mayor.
    else:
        print(f"{temp} Caluroso")
```

```python
# Recorremos los números del 1 al 100. (Usamos 101 para que el 100 quede incluido).
for numero in range(1, 101):

    # Verificamos si el número es divisible por 5.
    # El operador '%' nos da el resto de la división. Si es igual a 0, es divisible.
    if numero % 5 == 0:

        # Una vez que sabemos que es divisible por 5, revisamos si TAMBIÉN es divisible por 10.
        # Es clave hacer esta pregunta aquí adentro, porque todo múltiplo de 10 es múltiplo de 5.
        if numero % 10 == 0:
            print(f"{numero} Bip Bip")

        # Si el número es divisible por 5 pero falla la prueba del 10 (ej: 5, 15, 25)
        else:
            print(f"{numero} Bip")
```

```python
# Pedimos el primer número al usuario.
# Usamos int() para asegurarnos de que Python lo entienda como un número entero y no como texto.
numero1 = int(input("Por favor, ingresa el primer número: "))

# Pedimos el segundo número usando la misma lógica.
numero2 = int(input("Por favor, ingresa el segundo número: "))

# Realizamos la suma de las variables que guardaron los números
suma = numero1 + numero2

# Mostramos el resultado de la operación para darle feedback al usuario
print(f"\nLa suma de tus números es: {suma}")

# Evaluamos el resultado según las reglas pedidas
if suma < 50:
    print("Nivel principiante")

# Si llegó a este 'elif', la suma ya es 50 o mayor.
# Solo necesitamos comprobar que sea menor a 80.
elif suma < 80:
    print("Nivel intermedio")

# Si no es menor que 50, ni menor que 80, por descarte es 80 o más.
else:
    print("Nivel pro")
```

```python
# 1. Pedimos la edad y la convertimos a número entero similar al ejercicio anterior
edad = int(input("¿Cuál es tu edad?: "))

# 2. Pedimos los gustos musicales y de baile.
# Usamos .lower() al final para convertir cualquier respuesta a minúsculas automáticamente.
# Así, si escriben "SI" o "Si", la computadora lo lee como "si".
respuesta_musica = input("¿Te gusta la música? (si/no): ").lower()
respuesta_bailar = input("¿Te gusta bailar? (si/no): ").lower()

# 3. Convertimos las respuestas de texto a variables lógicas (Verdadero o Falso)
# Si la respuesta es "si" o "sí" (con tilde), la variable guardará un True (Verdadero).
le_gusta_musica = (respuesta_musica == "si" or respuesta_musica == "sí")
le_gusta_bailar = (respuesta_bailar == "si" or respuesta_bailar == "sí")

# 4. Evaluamos el perfil cruzando los datos
# Usamos 'and' para exigir que todas las condiciones se cumplan al mismo tiempo.
if edad >= 18 and le_gusta_musica and le_gusta_bailar:
    print("Eres una estrella de la fiesta")

# Usamos 'not' para comprobar si algo es Falso (en este caso, que NO le guste bailar)
elif le_gusta_musica and not le_gusta_bailar:
    print("Te gusta la música, pero no tanto bailar")

# Si no encaja en ninguno de los perfiles anteriores, cae aquí por descarte.
else:
    print("Tienes tu propio estilo")
```
